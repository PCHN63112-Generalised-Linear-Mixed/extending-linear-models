# The Analysis of Deviance
Last semester, we also saw how tests on multiple parameters can be conceptualised in terms of *model comparisons*. More specifically, we can compare a model containing the parameters of interest (the *full* model: $\mathcal{M}_{f}$) to a model that does *not* contain the parameters of interest (the *null* model: $\mathcal{M}_{0}$). The degree to which the error reduces between these models is equivalent to the sum-of-squares from the traditional ANOVA table. As such, the ANOVA is most easily conceptualised as a *model comparison procedure*.

A very similar perspective can be taken with GLMs. However, the process needs to be more general and applicable to data from non-normal distributions. To achieve this, the concept of the *residual sums-of-squares* (RSS) from the normal linear model is generalised to a related concept called *deviance*. This allows for a GLM version of the ANOVA, known as the *Analysis of Deviance*.


### Deviance


Deviance is based on comparing a model ($\mathcal{M}_{1}$) to a *saturated* version of the model ($\mathcal{M}_{sat}$) that has enough parameters to fit the observed data as closely as possible. In other words, deviance tells us how our model differ from the best possible job that the model could do. This has a direct connection to error in the normal linear model, where the residuals tell us how far the fitted values are from a perfect fit to the observed data.

As an example, we might have a Poisson model that assumes

$$
y_i \sim \text{Pois}(\lambda),
$$

such that there is a global $\lambda$ parameter and thus a single distribution from which all of the data arise. A saturated version of this would be

$$
y_i \sim \text{Pois}(\lambda_i),
$$

such that each datapoint has its own separate parameter. As there is no other information to constrain these parameters, $\hat{\lambda}_i = y_i$, making each observed value as likely as possible under a Poisson model. The deviance tells us how far our fitted model is from this *perfect* fit and is defined as

$$
    D_{{\mathcal{M}}_{1}} = 2\left(\ell_{sat} - \ell_{1}\right).
$$

Here, the $\ell$ symbol corresponds to the *log-likelihood*, which is defined as

$$
    \ell_{1} = \log\left[p(y|\mathcal{M}_{1})\right].
$$

In other words, this captures the probability of the data, assuming that model $\mathcal{M}_{1}$ is correct. As such, in simple terms, the deviance compares how well our model explains the data with how well a model that fits as well as possible explains the data.

### Why Do We Need Deviance?
To understand why deviance is needed in the GLM framework, we need to go back to the normal linear model and consider what the *residuals* tell us. Under the normal linear model, a residual is given as

$$
\epsilon_{i} = y_{i} - \hat{\mu}_{i} = \text{raw data} - \text{model prediction}.
$$

This is a *distance* measure which is equally meaningful irrespective of the predicted value. Now, consider two Poisson models, one with $y_{i} = 3$ and $\hat{\mu}_{i} = 1$, and one with $y_{i} = 102$ and $\hat{\mu}_{i} = 100$. In both cases, the residual is

$$
\epsilon_{i} = y_{i} - \hat{\mu}_{i} = 3 - 1 = 102 - 100 = 2. 
$$

However, this distance does not carry the same meaning in each case. Because the Poisson variance scales with the mean, a discrepancy of 2 with a mean of 1 is much more unexpected than a discrepancy of 2 with a mean of 100. As such, the *degree of error* is not the same. We therefore cannot simply use the raw residuals to capture the error because its meaning depends upon the distribution assumed *at that specific mean*. Under the normal linear model, the variance does not depend upon the mean, so the same-sized residual has the same meaning wherever it occurs. This is not true of many other distributions.

Deviance provides a solution to this problem. By moving from calculations based purely on distance to calculations based on *probability*, we can capture whether a particular discrepancy is considered large or small within the context of the assumed distribution. On its own, however, the probability of the data under the assumed model can be very small and difficult to interpret. Much in the same way that we would not simply inspect the predicted values from a normal linear model, we need something against which to compare it. The saturated model provides a natural zero-point because it represents the best fit to the observed data permitted by the assumed distribution.

The saturated model therefore plays a similar role to the raw data when calculating residuals in the normal linear model. A perfect fit is, by definition, indistinguishable from the raw data. As such, an alternative way of thinking about residuals is

$$
\begin{aligned}
    \epsilon_{i} &= \text{raw value} - \text{predicted value} \\
                 &\approx \text{perfect fit} - \text{current fit}
\end{aligned}
$$

making the deviance conceptually defined as

$$
D_{i} \approx \text{probability under perfect fit} - \text{probability under current fit}.
$$

This is not the literal formula for deviance, but it captures the core idea, as both residuals and deviance measure how far the current model is from a perfect fit. The difference is that residuals measure this using *distance*, whereas deviance measures it using *probability*.

:::{tip} The Role of the Saturated Model
In order to fully understand deviance, we need to understand the nature of the saturated model and a *perfect fit*. On the face of it, it might sound as though the saturated model is what we want, as it fits the observed data as well as possible. This would seem to imply that our aim should be a deviance of 0. However, this is not generally the aim.

A saturated model has enough flexibility to reproduce the observed data as closely as the assumed distribution allows. In doing so, however, it will capture not only the underlying signal in the data, but also the particular noise present in that sample. This is the problem of *over-fitting*. A model that is over-fit may describe the current dataset extremely well, but perform poorly when applied to a new sample.

Instead, we want a model that captures something about the underlying data-generating process that will generalise beyond the particular observations used to fit it. This usually requires accepting some discrepancy between the observed data and the fitted model.

The saturated model is therefore not an *aim*. It is a *baseline* representing the closest possible fit to the observed data. Deviance tells us how much the fitted model differs from that baseline after imposing a simpler structure that is intended to generalise beyond the current sample.
:::

### Comparing Deviance Betweeen Models
Taking all the above into account, we can treat deviance as the analogue to the residual sums-of-squares (RSS) in a linear model. As such, we can approach model comparison under the GLM in much the same way as under the normal linear model by calculating the *reduction in deviance* between two competing models. Under the normal linear model we might construct a test comparing a *full* model ($\mathcal{M}_{F}$) to a *reduced* model ($\mathcal{M}_{R}$) using

$$
    RSS_{R} - RSS_{F}
$$

and so, for a GLM, we can instead use

$$
    D_{R} - D_{F}.
$$

This then raises the possibility of construction an ANOVA-like table of omnibus tests from a GLM which, more correctly, is termed an *Analysis of Deviance*. 

In order to do this, we need to construct a test statistic from the change in deviance that is analogous to the ANOVA $F$-statistic, with a known null distribution. Luckily, it turns out that this subtraction of deviance is identical to an existing hypothesis testing framework known as the *likelihood ratio*.

### The Likelihood-Ratio Test
Within the framework of maximum likelihood estimation, two models can be compared through direct comparison of their likelihoods. This gives

$$
\Lambda = \frac{L_{R}}{L_{F}},
$$

where $L_R$ is the likelihood of the *reduced* model and $L_F$ is the likelihood of the *full* model. The magnitude of $\Lambda$ therefore indicates how well the data supports one model over the other. If $\Lambda = 1$ then the data supports both models equally, with $\Lambda < 1$ indicating that the data support $L_F$ more and $\Lambda > 1$ indicates that the data support $L_R$ more. 

As discussed last semester, in the framework of maximum likelihood it can be difficult to calculate the likelihood of a whole dataset because this involves multiplying many very small probabilities. As such, it is more usual to use the *log* of the likelihood, as this places very small likelihood values on a more manageable scale and turns multiplication into summation. This turns the *ratio of likelihoods* into the subtraction of *log-likelihoods* 

$$
\log(\Lambda) = \log\left(\frac{L_R}{L_F}\right) = \log(L_R) - \log(L_F) = \ell_{R} - \ell_{F}.
$$

As a final step, a result from mathematical statistics known as [Wilks' Theorem](https://en.wikipedia.org/wiki/Wilks%27_theorem) shows that multiplying the log likelihood ratio by -2 leads to a statistic that is asymptotically distributed under the null as $\chi^{2}_{q}$, where $q$ is the difference in the number of parameters between the models. This provides the mechanism for calculating an asymptotic $p$-value from the likelihood-ratio test.

To connect this with deviance, note that the likelihood-ratio statistic is

$$
-2\log(\Lambda) = 2(\ell_{F} - \ell_{R}).
$$

Subtracting the deviance of a full model from a reduced model causes the saturated model to cancel, leading to

$$
D_R - D_F = 2\left[(\ell_{sat}-\ell_R) - (\ell_{sat}-\ell_F)\right] = 2(\ell_{F} - \ell_{R})
$$

which is *identical* to the likelihood ratio statistic. As such, subtraction of deviance *is* a likelihood ratio test, meaning we can calculate a $p$-value using its asymptotic $\chi^{2}_{q}$ distribution. This provides the final piece of the puzzle that allows for the full generalisation of an *Analysis of Variance* to an *Analysis of Deviance*.

```{tip} Key Point
Deviance provides a natural analogue to the residual sums-of-squares from the linear model. Two models can therefore be compared by comparison of their deviance. This provides a mechanism for defining an Analysis of Deviance, which is simply an ANOVA within the framework of the GLM. In order to do so, we need $p$-values and confidence intervals from comparisons of deviance. As it turns out, comparisons of deviance are identical to the likelihood ratio statistic and thus $p$-values can be calculated using the asymptotic $\chi^{2}_{q}$ distribution. Unfortunately, this has the same limitations as the asymptotic results used for inference on individual parameters with the Wald tests, meaning caution must be exercised with small samples.
```

:::{tip .simple icon=false} What You Now Know
In this section, we have discussed how an analogue to the ANOVA can be created for GLMs using the concept of deviance. After reading this section, you should have a good sense of:
- The idea that deviance is a generalisation of the residual sum-of-squares (RSS) from the standard linear model.
- That deviance is defined as a comparison between the current model and a *saturated* version that fits the data as well as possible.
- The idea that comparisons of deviance across nested models provides a similar idea to comparisons of RSS in an ANOVA, allowing the same model comparison perspective to be used. This is known as an *Analysis of Deviance*.
- The result that comparisons of deviance are identical to a likelihood-ratio test comparing two nested models, meaning that the deviance statistic is *asymptotically* distributed as a $\chi^{2}_{q}$ under the null.
- The general limitation that, like the Wald test, the deviance tests are *asymptotic* in terms of their $p$-values and thus are less trustworthy under small samples.
:::
