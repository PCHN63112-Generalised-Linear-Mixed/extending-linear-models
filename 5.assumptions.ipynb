# Model Assumptions
As we saw with the normal linear model, once the model has been estimated we need to assess whether the assumptions are met before proceeding to inference. Under the normal linear model, we only had a single set of assumptions. Under a GLM, we have a set of assumptions that are generic across GLMs, but we also need to consider these within the context of the specific GLM we are using. As such, not every diagnostic is going to be informative or even relevant for every GLM.

## Generic GLM Assumptions
As discussed above, all GLMs are based on three core aspects:

$$
\begin{align*}
    E(y_{i})   &= \mu_{i} \\
    \mu_{i} &= g^{-1}\left(\beta_{0} + \beta_{1}x_{i1} + \dots + \beta_{k}x_{ik}\right) \\
    \text{Var}(y_{i}) &= \phi V(\mu_{i})
\end{align*}
$$

This leads to several key assumptions about the data:

1. The observations are independent of each other (no correlation).
2. The observations follow the assumed conditional distribution of the outcome variable.
3. The mean is accurately described using the (transformed) regression model.
4. The conditional variance follows the form for the assumed distribution.

These are all quite generic, so the trick is understanding how they apply in different cases, particular those when some of the assumptions either simplify or become trivial. We will see examples of interpreting these in the context of different GLMs over the next few weeks.

## Checking Assumptions with Residuals
Much like with the normal linear model, the assumptions of GLMs can be checked by producing plots. This includes effects plots of the fitted model and plots of the residuals. However, the nature of residuals changes within the GLM framework because there are *multiple* ways that residuals can be defined. This is a complication of the variety of distributions available. In addition, although a standard set of residual plots can be created, these won't all be equally useful or relevant for every type of GLM. As such, the specific type of model needs to be taken into account, as we will see as we progress. For the moment, we will simply define some of the residuals we will come across in the next few weeks.

### Response Residuals
The first type of residuals are known as *response* residuals and are the closest to the definition of residuals within the normal linear model, but are the least useful for diagnostic purposes. These are simply defined as

$$
    r_{i}^{(R)} = y_{i} - \hat{\mu}_{i},
$$

where $\hat{\mu}_{i}$ is the expected value of $y_{i}$, given the model. In the case of binary logistic regression, $\hat{\mu}_{i} = \hat{\pi}_{i}$, meaning that the predicted probability of success for obervation $i$ is subtracted from the raw data. The higher the probability of success, the smaller $r_{i}^{(R)}$ will be for values of 1 and the bigger $r_{i}^{(R)}$ will be for values of 0. So, these values do provide some indication of how close the prediction is for each value of the outcome.

Unfortunately, response residuals are less useful for checking assumptions because they do not accommodate the fact that many distributions used with GLMs do not have constant variance. In many cases, the variance is a function of the value of the mean. Because response residuals ignore this, they can demonstrate patterns that may suggest a problem when in reality this is a simply a reflection of the way different distributions are defined. 

### Pearson Residuals
The second type of residuals are known as *Pearson* residuls and are designed to fix the problem indicated above by scaling the response residuals by the variance function of each data point. They are defined as

$$
    r_{i}^{(P)} = \frac{y_{i} - \hat{\mu}_{i}}{V(\hat{\mu}_{i})},
$$

effectively removing any difference in variance from the residuals and making the spread of residuals constant across the data points. However, we have the same problem with residuals that we discussed last semester. Namely, that the *residuals* cannot be truly homoscedastic, even with the correction above, due to *leverage*. Much in the same way that we can standardise residuals from the normal linear model, we can do the same with Pearson residuals to accommodate leverage and place all residuals on the same scale. This follows an adaption of the formula from last semester to give

$$
    r_{i}^{(Pz)} = \frac{y_{i} - \hat{\mu}_{i}}{\sqrt{\phi V(\hat{\mu}_{i})(1-h_{i})}},
$$

where $h_{i}$ is the leverage for observation $i$.

### Deviance Residuals
As a final alternative to Pearson residuals, there are also residuals based on the *deviance*. As mentioned earlier, deviance can provided a probability-based account of model fit and thus provides an opportunity for a definition of residuals that applies across all GLMs. Furthermore, because the deviance is based upon the likelihood, the variance function already accommodated into its definition. 

The deviance residual for observation $i$ is defined as

$$
    r_{i}^{(D)} = \text{sign}(y_{i} - \hat{\mu}_{i}) \sqrt{D_{i}},
$$

where $\text{sign}(y_{i} - \hat{\mu}_{i})$ simply makes $r_{i}^{(D)}$ positive or negative, depending upon whether $y_{i} - \hat{\mu}_{i}$ is positive or negative. What is particularly advantageous here is that $\sum{r_{i}^{2}} = D$, providing a similar connection between deviance residuals and the model deviance as we have in the normal linear model between residuals and the model variance. The deviance residuals can also be standardised using

$$
    r_{i}^{(Dz)} = \frac{r_{i}^{(D)}}{\sqrt{\phi (1-h_{i})}}
$$

which, for some GLMs, will approximate a standard normal distribution, allowing assessment of model fit using similar diagnostics as used for the normal linear model. However, the usefulness of this will depend upon the specific GLM in question.
