# Extending Linear Models Beyond the Normal Distribution
Last semester, we spent a lot of time exploring the normal linear model. While this is a highly-flexible tool for analysing data from a variety of experiments with different forms of predictor variables, there was one clear limitation of this approach. While the predictors could either be continuous or categorical, the normal linear model required the outcome variable to be continuous. Furthermore, it was assumed that, conditional on the value of the predictors, the outcome was also normally distributed. 

Whilst applicable to lots of experiments, there are clear cases where the outcome is neither continuous, nor can realistically be modelled using a normal distribution. As such, we either restrict all experiments to only measuring continuous phenomena that can be considered normally distributed, or we need a more general approach.

This lesson introduces the *Generalised Linear Models* framework, which is designed to allow regression models to be applied to data from a variety of different distributions. This allows extension of the normal linear model to cases where the outcome variable is binary, nominal, ordinal, a count or skewed. This provides a very powerful extension to our existing framework, increasing the potential for designing and analysing experiments using a variety of different outcome measurements.

Over the coming weeks, we will see many examples of applying GLMs to different types of data. In this lesson, we will focus on motivating the need for GLMs, as well as presenting some general principles we will need going forward. Because of this, the lesson is quite generic and may need to be revisited once some applied examples have been seen later in the unit.
